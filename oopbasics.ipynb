```python
class employee:
    #default constructor
    def __init__(self):
        self.empid = 1001
        self.empname = "jhon"
        self.empsalary = 23253476.4634

    #parameterized constructor
    def __init__(self, empid, empname, empsalary):
        self.empid = empid
        self.empname = empname
        self.empsalary = empsalary

# behavior or methods can be mentioned here
    def showempdetails(self):
        print("Employee ID:", self.empid , "Employee Name:", self.empname, "Employee Salary:", self.empsalary)
```

```python
#create object of employee class

empobj1= employee(1001, "jhon", 23253476.4634)
empobj1.showempdetails()
```

```python
class Account:
    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print("Deposit successful. New balance:", self.balance)

    def withdraw(self, amount):
        self.withdrawal_amount = amount
    def display_account_info(self):
        print("Account Number:", self.account_number,"Deposit :", self.balance, "Account Holder:", self.account_holder)
      
```

```python
#create obj of account class
accobj1=Account(1001, "jhon", 23253476.4634)
accobj1.display_account_info()
```

```python
class Manager(employee):
    def __init__(self, empid, empname, empsalary, perks):
        super().__init__(empid, empname, empsalary)
        self.perks = perks

    def showtotalsalary(self):
        total_salary = self.empsalary + self.perks
        print("Total Salary : ", total_salary)  # "Manager ID:", self.empid, "Manager Name:", self.empname, "Total Salary:", total_salary)
```

```python
mgrobj = Manager(1002, "Alice", 200000.00, 1000.00)
mgrobj.showempdetails()
mgrobj.showtotalsalary()
```
